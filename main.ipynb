```python
def compute_cost(x,y,w,b):
    m = x.shape[0]
    cost = 0

    for i in range(m):
        f_wb = w * x[i] + b
        cost += (y[i] - f_wb)**2
    
    return cost/(2*m)
```

```python
def compute_gradient(x,y,w,b):
    m = x.shape[0]

    dj_dw = 0
    dj_db = 0

    for i in range(m):
        f_wb = (x[i]*w+b)
        error = f_wb - y[i] 
        dj_dw += error * x[i]
        dj_db += error
    
    return (dj_dw/m, dj_db/m)
```
